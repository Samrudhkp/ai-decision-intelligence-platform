````python
from pathlib import Path
import pandas as pd

from src.utils.data_generation import write_datasets

paths = write_datasets()
churn = pd.read_csv(paths["churn"])
revenue = pd.read_csv(paths["revenue"], parse_dates=["month"])
churn.head(), revenue.head()
````

````python
print("Churn rate:", churn["churned"].mean().round(3))
churn.groupby("contract_type")["churned"].mean().sort_values(ascending=False)
````
